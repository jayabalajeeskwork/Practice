```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("employee_salary_dataset.csv")
df.head()
```

```python
plt.plot(df['Experience_Years'], df['Monthly_Salary'])
plt.show()
```

```python
plt.plot(
    df['Experience_Years'],
    df['Monthly_Salary'],
    color='red',
    linestyle='--',
    marker='o',
    linewidth=3
)

plt.title("Salary Growth")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

```python
plt.bar(
    df['Experience_Years'],
    df['Monthly_Salary']
)

plt.show()
```

```python
plt.bar(
    df['Experience_Years'],
    df['Monthly_Salary'],
    color='green'
)

plt.title("Experience vs Salary")
plt.show()
```

```python
plt.barh(
    df['Experience_Years'],
    df['Monthly_Salary'],
    color='orange'
)

plt.show()
```

```python
salary = [30,40,50]
labels = ['A','B','C']

plt.pie(
    salary,
    labels=labels,
    autopct='%1.1f%%'
)

plt.show()
```

```python
plt.scatter(
    df['Experience_Years'],
    df['Monthly_Salary'],
    color='purple'
)

plt.show()
```

```python
plt.boxplot(df['Monthly_Salary'])
plt.show()
```

```python
plt.fill_between(
    df['Experience_Years'],
    df['Monthly_Salary']
)

plt.show()
```

```python
x=[1,2,3,4]

y1=[10,20,30,40]
y2=[15,25,35,45]

plt.plot(x,y1,label='Sales')
plt.plot(x,y2,label='Profit')

plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employee_salary_dataset.csv")

plt.figure(figsize=(10,5))

plt.plot(
    df['Experience_Years'],
    df['Monthly_Salary'],
    color='blue',
    marker='o',
    linestyle='--',
    linewidth=3
)

plt.title("Salary Analysis")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)

plt.show()
```
